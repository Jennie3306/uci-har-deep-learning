```python
import numpy as np, pandas as pd
from pathlib import Path
from sklearn.model_selection import GroupKFold
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, accuracy_score
from har_pytorch import download_dataset

# Same data order as har_pytorch.py (train then test) -> same GroupKFold folds
root = download_dataset(Path("data"))
X = np.vstack([np.loadtxt(root / s / f"X_{s}.txt") for s in ("train", "test")])
y = np.concatenate([np.loadtxt(root / s / f"y_{s}.txt", dtype=int) for s in ("train", "test")])
g = np.concatenate([np.loadtxt(root / s / f"subject_{s}.txt", dtype=int) for s in ("train", "test")])
print(X.shape, len(np.unique(g)), "subjects")

models = {
    "LogisticRegression": make_pipeline(StandardScaler(), LogisticRegression(max_iter=3000)),
    "LinearSVC": make_pipeline(StandardScaler(), LinearSVC(C=1.0, max_iter=5000)),
    "RandomForest": RandomForestClassifier(n_estimators=300, n_jobs=-1, random_state=42),
}

rows = []
for name, model in models.items():
    f1s, accs = [], []
    for fold, (tr, te) in enumerate(GroupKFold(n_splits=5).split(X, y, g), start=1):
        if name == "LogisticRegression":
            print(f"Fold {fold}: test subjects {sorted(set(g[te].tolist()))}")
        model.fit(X[tr], y[tr])
        pred = model.predict(X[te])
        f1s.append(f1_score(y[te], pred, average="macro"))
        accs.append(accuracy_score(y[te], pred))
    rows.append({
        "model": name,
        "macro_f1_mean": np.mean(f1s),
        "macro_f1_std": np.std(f1s, ddof=1),
        "accuracy_mean": np.mean(accs),
        "fold_macro_f1": " ".join(f"{v:.4f}" for v in f1s),
    })
    print(f"{name}: macro-F1 {np.mean(f1s):.4f} ± {np.std(f1s, ddof=1):.4f}")

df = pd.DataFrame(rows).round(4)
df.to_csv("results/sklearn_baseline.csv", index=False)
df
```
